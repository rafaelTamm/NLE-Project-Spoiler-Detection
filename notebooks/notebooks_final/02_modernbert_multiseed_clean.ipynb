# ModernBERT Spoiler Detection

This notebook fine-tunes ModernBERT for spoiler detection on IMDb movie reviews and Goodreads book reviews.

We evaluate the model in four settings:

1. IMDb → IMDb (in-domain)
2. IMDb → Goodreads (cross-domain)
3. Goodreads → Goodreads (in-domain)
4. Goodreads → IMDb (cross-domain)

The same train, validation, and test splits as in the TF-IDF baseline are used to ensure a direct comparison between both approaches.

```python
import os
import pandas as pd
import numpy as np
import torch

from datasets import Dataset, DatasetDict
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    DataCollatorWithPadding,
    TrainingArguments,
    Trainer,
    set_seed
)
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

## Load Data


```python
imdb_train = pd.read_csv("data/imdb_train.csv")
imdb_val = pd.read_csv("data/imdb_val.csv")
imdb_test = pd.read_csv("data/imdb_test.csv")

goodreads_train = pd.read_csv("data/goodreads_train.csv")
goodreads_val = pd.read_csv("data/goodreads_val.csv")
goodreads_test = pd.read_csv("data/goodreads_test.csv")

datasets = {
    "IMDb train": imdb_train,
    "IMDb val": imdb_val,
    "IMDb test": imdb_test,
    "Goodreads train": goodreads_train,
    "Goodreads val": goodreads_val,
    "Goodreads test": goodreads_test,
}

for name, df in datasets.items():
    print(f"{name}: {len(df)} reviews")
```

```python
datasets = {
    "IMDb Train": imdb_train,
    "IMDb Validation": imdb_val,
    "IMDb Test": imdb_test,
    "Goodreads Train": goodreads_train,
    "Goodreads Validation": goodreads_val,
    "Goodreads Test": goodreads_test,
}

for name, df in datasets.items():
    print(name)
    print("Reviews:", len(df))
    print("Items:", df["item_id"].nunique())
    print("Spoiler %:", round(df["is_spoiler"].astype(int).mean() * 100, 2))
    print()
```

```python
def check_item_overlap(train, val, test, dataset_name):
    train_items = set(train["item_id"])
    val_items = set(val["item_id"])
    test_items = set(test["item_id"])

    print(f"{dataset_name}:")
    print("Train / Validation overlap:", len(train_items & val_items))
    print("Train / Test overlap:", len(train_items & test_items))
    print("Validation / Test overlap:", len(val_items & test_items))
    print()


check_item_overlap(imdb_train, imdb_val, imdb_test, "IMDb")
check_item_overlap(goodreads_train, goodreads_val, goodreads_test, "Goodreads")
```

## Prepare Hugging Face Datasets


```python
imdb_dataset = DatasetDict({
    "train": Dataset.from_pandas(imdb_train),
    "validation": Dataset.from_pandas(imdb_val),
    "test": Dataset.from_pandas(imdb_test)
})

goodreads_dataset = DatasetDict({
    "train": Dataset.from_pandas(goodreads_train),
    "validation": Dataset.from_pandas(goodreads_val),
    "test": Dataset.from_pandas(goodreads_test)
})
```

## Prepare Labels


```python
def prepare_labels(example):
    example["label"] = int(example["is_spoiler"])
    return example

imdb_dataset = imdb_dataset.map(prepare_labels)
goodreads_dataset = goodreads_dataset.map(prepare_labels)

print("IMDb:", imdb_dataset["train"][0]["is_spoiler"], "->", imdb_dataset["train"][0]["label"])
print("Goodreads:", goodreads_dataset["train"][0]["is_spoiler"], "->", goodreads_dataset["train"][0]["label"])
```

## Dataset Split Statistics

Since the datasets were split by item ID using GroupShuffleSplit, we inspect
the size, number of unique items, and class distribution of each split.
This ensures that potential differences in class distribution between
training, validation, and test sets are documented.

```python
def get_split_statistics(dataset, dataset_name):
    rows = []

    for split_name in ["train", "validation", "test"]:
        split = dataset[split_name]

        labels = pd.Series(split["label"])
        spoiler_count = (labels == 1).sum()
        non_spoiler_count = (labels == 0).sum()

        rows.append({
            "Dataset": dataset_name,
            "Split": split_name,
            "Reviews": len(split),
            "Unique Items": len(set(split["item_id"])),
            "Spoilers": spoiler_count,
            "Non-Spoilers": non_spoiler_count,
            "Spoiler %": spoiler_count / len(split) * 100,
            "Non-Spoiler %": non_spoiler_count / len(split) * 100
        })

    return pd.DataFrame(rows)


split_statistics = pd.concat([
    get_split_statistics(imdb_dataset, "IMDb"),
    get_split_statistics(goodreads_dataset, "Goodreads")
], ignore_index=True)

display(split_statistics.round(2))
```

## ModernBERT Tokenizer

The previous length analysis showed that a maximum sequence length of 1024 preserves substantially more spoiler-review content than 512 while keeping training feasible.


```python
checkpoint = "answerdotai/ModernBERT-base"
tokenizer = AutoTokenizer.from_pretrained(checkpoint)
```

```python
def analyze_token_lengths(dataset, dataset_name):
    all_texts = []

    for split in ["train", "validation", "test"]:
        all_texts.extend(dataset[split]["review_text"])

    token_lengths = [
        len(tokenizer(text, truncation=False)["input_ids"])
        for text in all_texts
    ]

    print(f"{dataset_name}")
    print(f"Average length: {np.mean(token_lengths):.1f}")
    print(f"Median length: {np.median(token_lengths):.1f}")
    print(f"90th percentile: {np.percentile(token_lengths, 90):.1f}")
    print(f"95th percentile: {np.percentile(token_lengths, 95):.1f}")
    print(f"99th percentile: {np.percentile(token_lengths, 99):.1f}")
    print(f"Maximum length: {np.max(token_lengths)}")
    print(f"Over 512 tokens: {np.mean(np.array(token_lengths) > 512) * 100:.1f}%")
    print(f"Over 1024 tokens: {np.mean(np.array(token_lengths) > 1024) * 100:.1f}%")
    print()


analyze_token_lengths(imdb_dataset, "IMDb")
analyze_token_lengths(goodreads_dataset, "Goodreads")
```

## Tokenization


```python
MAX_LENGTH = 1024

def tokenize_function(examples):
    return tokenizer(
        examples["review_text"],
        truncation=True,
        max_length=MAX_LENGTH
    )

tokenized_imdb = imdb_dataset.map(tokenize_function, batched=True)
tokenized_goodreads = goodreads_dataset.map(tokenize_function, batched=True)

print(tokenized_imdb["train"][0].keys())
print("IMDb token count:", len(tokenized_imdb["train"][0]["input_ids"]))
print("Goodreads token count:", len(tokenized_goodreads["train"][0]["input_ids"]))
```

## Dynamic Padding


```python
data_collator = DataCollatorWithPadding(tokenizer=tokenizer)
```

## Evaluation Metrics


```python
def compute_metrics(eval_pred):
    predictions, labels = eval_pred
    predictions = np.argmax(predictions, axis=-1)

    return {
        "accuracy": accuracy_score(labels, predictions),
        "precision": precision_score(labels, predictions, zero_division=0),
        "recall": recall_score(labels, predictions, zero_division=0),
        "f1": f1_score(labels, predictions, zero_division=0),
        "macro_f1": f1_score(labels, predictions, average="macro", zero_division=0)
    }
```

## Multi-Seed Configuration

The original data sampling and train/validation/test splits remain fixed with seed 42.
Only ModernBERT fine-tuning is repeated with three training seeds: 42, 123, and 456.

All other training hyperparameters are kept identical across seeds.
Checkpoints are stored separately for every source domain and seed.


```python
from google.colab import drive
drive.mount("/content/drive")

MULTISEED_DIR = "/content/drive/MyDrive/NLP_Spoiler_Project/multiseed"
os.makedirs(MULTISEED_DIR, exist_ok=True)

SEEDS = [42, 123, 456]
SEEDS_TO_TRAIN = [456]

print("Seeds used in the experiment:", SEEDS)
print("New seeds to train:", SEEDS_TO_TRAIN)
print("Checkpoint directory:", MULTISEED_DIR)
```

## Split Sanity Check

The dataset partitions are not regenerated for the multi-seed experiment.
We only verify again that item IDs do not overlap between train, validation, and test.


```python
def check_split_overlap(train_df, val_df, test_df, name):
    train_ids = set(train_df["item_id"])
    val_ids = set(val_df["item_id"])
    test_ids = set(test_df["item_id"])

    print(name)
    print("Train/Validation overlap:", len(train_ids & val_ids))
    print("Train/Test overlap:", len(train_ids & test_ids))
    print("Validation/Test overlap:", len(val_ids & test_ids))
    print()

check_split_overlap(imdb_train, imdb_val, imdb_test, "IMDb")
check_split_overlap(goodreads_train, goodreads_val, goodreads_test, "Goodreads")
```

## IMDb Multi-Seed Fine-Tuning

For each seed, a fresh `answerdotai/ModernBERT-base` sequence-classification model is initialized.
The model is selected exclusively by IMDb validation Macro F1 and then evaluated on both IMDb and Goodreads test data.


```python
imdb_seed_results = []

for seed in SEEDS_TO_TRAIN:
    print(f"\n{'=' * 70}")
    print(f"IMDb training — seed {seed}")
    print(f"{'=' * 70}")

    set_seed(seed)

    imdb_output_dir = os.path.join(
        MULTISEED_DIR,
        f"imdb_seed_{seed}"
    )
    os.makedirs(imdb_output_dir, exist_ok=True)

    imdb_model = AutoModelForSequenceClassification.from_pretrained(
        checkpoint,
        num_labels=2
    )

    imdb_training_args = TrainingArguments(
        output_dir=imdb_output_dir,
        num_train_epochs=3,
        learning_rate=2e-5,
        per_device_train_batch_size=2,
        per_device_eval_batch_size=4,
        gradient_accumulation_steps=4,
        eval_strategy="epoch",
        save_strategy="epoch",
        load_best_model_at_end=True,
        metric_for_best_model="macro_f1",
        greater_is_better=True,
        logging_steps=100,
        save_total_limit=2,
        seed=seed,
        report_to="none"
    )

    imdb_trainer = Trainer(
        model=imdb_model,
        args=imdb_training_args,
        train_dataset=tokenized_imdb["train"],
        eval_dataset=tokenized_imdb["validation"],
        data_collator=data_collator,
        compute_metrics=compute_metrics
    )

    imdb_trainer.train()

    imdb_in_domain = imdb_trainer.evaluate(tokenized_imdb["test"])
    imdb_to_goodreads = imdb_trainer.evaluate(tokenized_goodreads["test"])

    imdb_seed_results.extend([
        {
            "Model": "ModernBERT",
            "Seed": seed,
            "Experiment": "IMDb → IMDb",
            "Accuracy": imdb_in_domain["eval_accuracy"],
            "Precision": imdb_in_domain["eval_precision"],
            "Recall": imdb_in_domain["eval_recall"],
            "F1": imdb_in_domain["eval_f1"],
            "Macro F1": imdb_in_domain["eval_macro_f1"],
        },
        {
            "Model": "ModernBERT",
            "Seed": seed,
            "Experiment": "IMDb → Goodreads",
            "Accuracy": imdb_to_goodreads["eval_accuracy"],
            "Precision": imdb_to_goodreads["eval_precision"],
            "Recall": imdb_to_goodreads["eval_recall"],
            "F1": imdb_to_goodreads["eval_f1"],
            "Macro F1": imdb_to_goodreads["eval_macro_f1"],
        }
    ])

    print(f"Finished IMDb seed {seed}")
    print("IMDb → IMDb Macro F1:", imdb_in_domain["eval_macro_f1"])
    print("IMDb → Goodreads Macro F1:", imdb_to_goodreads["eval_macro_f1"])
```

```python
seed_123_results = pd.DataFrame(imdb_seed_results)

seed_123_results.to_csv(
    "/content/drive/MyDrive/NLP_Spoiler_Project/multiseed/imdb_seed_123_results.csv",
    index=False
)

display(seed_123_results)
print("IMDb seed 123 results saved to Drive.")
```

## Goodreads Multi-Seed Fine-Tuning

For each seed, a fresh `answerdotai/ModernBERT-base` sequence-classification model is initialized.
The model is selected exclusively by Goodreads validation Macro F1 and then evaluated on both Goodreads and IMDb test data.


```python
goodreads_seed_results = []

for seed in SEEDS_TO_TRAIN:
    print(f"\n{'=' * 70}")
    print(f"Goodreads training — seed {seed}")
    print(f"{'=' * 70}")

    set_seed(seed)

    goodreads_output_dir = os.path.join(
        MULTISEED_DIR,
        f"goodreads_seed_{seed}"
    )
    os.makedirs(goodreads_output_dir, exist_ok=True)

    goodreads_model = AutoModelForSequenceClassification.from_pretrained(
        checkpoint,
        num_labels=2
    )

    goodreads_training_args = TrainingArguments(
        output_dir=goodreads_output_dir,
        num_train_epochs=3,
        learning_rate=2e-5,
        per_device_train_batch_size=2,
        per_device_eval_batch_size=4,
        gradient_accumulation_steps=4,
        eval_strategy="epoch",
        save_strategy="epoch",
        load_best_model_at_end=True,
        metric_for_best_model="macro_f1",
        greater_is_better=True,
        logging_steps=100,
        save_total_limit=2,
        seed=seed,
        report_to="none"
    )

    goodreads_trainer = Trainer(
        model=goodreads_model,
        args=goodreads_training_args,
        train_dataset=tokenized_goodreads["train"],
        eval_dataset=tokenized_goodreads["validation"],
        data_collator=data_collator,
        compute_metrics=compute_metrics
    )

    goodreads_trainer.train()

    goodreads_in_domain = goodreads_trainer.evaluate(tokenized_goodreads["test"])
    goodreads_to_imdb = goodreads_trainer.evaluate(tokenized_imdb["test"])

    goodreads_seed_results.extend([
        {
            "Model": "ModernBERT",
            "Seed": seed,
            "Experiment": "Goodreads → Goodreads",
            "Accuracy": goodreads_in_domain["eval_accuracy"],
            "Precision": goodreads_in_domain["eval_precision"],
            "Recall": goodreads_in_domain["eval_recall"],
            "F1": goodreads_in_domain["eval_f1"],
            "Macro F1": goodreads_in_domain["eval_macro_f1"],
        },
        {
            "Model": "ModernBERT",
            "Seed": seed,
            "Experiment": "Goodreads → IMDb",
            "Accuracy": goodreads_to_imdb["eval_accuracy"],
            "Precision": goodreads_to_imdb["eval_precision"],
            "Recall": goodreads_to_imdb["eval_recall"],
            "F1": goodreads_to_imdb["eval_f1"],
            "Macro F1": goodreads_to_imdb["eval_macro_f1"],
        }
    ])

    print(f"Finished Goodreads seed {seed}")
    print("Goodreads → Goodreads Macro F1:", goodreads_in_domain["eval_macro_f1"])
    print("Goodreads → IMDb Macro F1:", goodreads_to_imdb["eval_macro_f1"])
```
